# GNN Depth vs. Robust Expressivity: Oversmoothing Study

Supports **GIN** and **GCN** — set `ARCH` in the Setup cell.

- **GIN** (Xu et al. 2019): injective sum aggregation, maximally expressive.
- **GCN** (Kipf & Welling 2017): normalised mean aggregation, prone to oversmoothing.

Evaluated across four random graph families (`GRAPH_TYPES` in the Setup cell), so oversmoothing trends are checked for robustness to graph structure, not just Erdos-Renyi:

- **ER** — Erdos-Renyi (iid edges)
- **RGG** — random geometric graph (edges from spatial proximity)
- **BA** — Barabasi-Albert (preferential attachment / hubs)
- **WS** — Watts-Strogatz (small-world, ring lattice + rewiring)

**Core question:** Does increasing depth $L$ collapse the model-based margin $\hat{\gamma}_{f(A)}$ through oversmoothing, and does this directly predict alignment failure under edge perturbations?

$$\hat{\gamma}_{f(A)} = \min_{i \neq j} \|[f(A)]_i - [f(A)]_j\|_2$$

Robust node-level expressivity (Proposition 1): the GNN survives perturbation $k$ if $\|f(A) - f(A')\|_2 < \hat{\gamma}_{f(A)}$.

## 0. Setup

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.nn import GINConv, GCNConv
from torch_geometric.data import Data

import numpy as np
import networkx as nx
from scipy.optimize import linear_sum_assignment
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import random, warnings
warnings.filterwarnings('ignore')

DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'Using device: {DEVICE}')

# ── Architecture choice ───────────────────────────────────────────────────────
ARCH = 'GIN'   # 'GIN' or 'GCN'

# ── Node feature choice ──────────────────────────────────────────────────────
# 'degree'       : node feature = degree of the *input* graph (recomputed from
#                  the perturbed graph at inference time, not the clean graph).
# 'all_ones'     : constant all-ones feature vector, invariant to perturbation.
# 'random_fixed' : a random vector drawn once per node when the clean graph is
#                  sampled; held fixed (acts as a node id) under edge perturbation.
FEATURE_TYPE = 'all_ones'   # 'degree' | 'all_ones' | 'random_fixed'

# ── Graph family choice ───────────────────────────────────────────────────────
# The full depth sweep is repeated per family so oversmoothing can be compared
# across structurally different random graph models, not just Erdos-Renyi.
GRAPH_TYPES = ['ER', 'RGG', 'BA', 'WS']
# ER  : Erdos-Renyi          — edge iid with prob. ER_PROB
# RGG : Random geometric     — points in [0,1]^2, edge if distance < RGG_RADIUS
# BA  : Barabasi-Albert      — preferential attachment, BA_M edges per new node
# WS  : Watts-Strogatz       — ring lattice (degree WS_K) rewired w.p. WS_P
GRAPH_COLORS = {'ER': 'tab:blue', 'RGG': 'tab:green', 'BA': 'tab:red', 'WS': 'tab:purple'}

# ── Hyperparameters ──────────────────────────────────────────────────────────
N_NODES      = 30
ER_PROB      = 0.4
RGG_RADIUS   = 0.35   # ≈ same expected mean degree as ER_PROB (~12)
BA_M         = 6      # ≈ same expected mean degree as ER_PROB (~12)
WS_K         = 12     # ≈ same expected mean degree as ER_PROB (~12)
WS_P         = 0.1
FEAT_DIM     = 1
HIDDEN_DIM   = 64
EMBED_DIM    = 32
TRAIN_STEPS  = 1000
LR           = 5e-4
N_TEST       = 80
K_VALUES     = [0, 1, 2, 3, 4, 5, 6, 8]
FIXED_MARGIN = 0.5
DEPTH_VALUES = [3,4,5,6]

plt.rcParams.update({'font.size': 11, 'axes.spines.top': False,
                     'axes.spines.right': False, 'figure.dpi': 130})
CMAP_L = cm.viridis
```

## 1. Graph Generation and Perturbation

```python
def compute_features(A, feat_dim, feature_type, rng):
    n = A.shape[0]
    if feature_type == 'degree':
        # degree of the graph actually passed in (clean or perturbed)
        degree = (A.sum(axis=1, keepdims=True) / n).astype(np.float32)
        if feat_dim > 1:
            degree = np.concatenate(
                [degree, np.zeros((n, feat_dim - 1), dtype=np.float32)], axis=1)
        return degree
    elif feature_type == 'all_ones':
        return np.ones((n, feat_dim), dtype=np.float32)
    elif feature_type == 'random_fixed':
        return rng.randn(n, feat_dim).astype(np.float32)
    else:
        raise ValueError(f'Unknown FEATURE_TYPE: {feature_type}')


def sample_er_graph(n=N_NODES, p=ER_PROB, feat_dim=FEAT_DIM, seed=None,
                    feature_type=FEATURE_TYPE):
    rng = np.random.RandomState(seed) if seed is not None else np.random
    A = np.zeros((n, n), dtype=np.float32)
    for i in range(n):
        for j in range(i + 1, n):
            if rng.rand() < p:
                A[i, j] = A[j, i] = 1.0
    feats = compute_features(A, feat_dim, feature_type, rng)
    return A, feats


def sample_rgg_graph(n=N_NODES, radius=RGG_RADIUS, feat_dim=FEAT_DIM, seed=None,
                     feature_type=FEATURE_TYPE):
    # random geometric graph: n points uniform in [0,1]^2, edge iff distance < radius
    G = nx.random_geometric_graph(n, radius, seed=int(seed) if seed is not None else None)
    A = nx.to_numpy_array(G, nodelist=range(n)).astype(np.float32)
    rng   = np.random.RandomState(seed) if seed is not None else np.random
    feats = compute_features(A, feat_dim, feature_type, rng)
    return A, feats


def sample_ba_graph(n=N_NODES, m=BA_M, feat_dim=FEAT_DIM, seed=None,
                    feature_type=FEATURE_TYPE):
    # Barabasi-Albert preferential attachment: each new node attaches to m existing nodes
    G = nx.barabasi_albert_graph(n, m, seed=int(seed) if seed is not None else None)
    A = nx.to_numpy_array(G, nodelist=range(n)).astype(np.float32)
    rng   = np.random.RandomState(seed) if seed is not None else np.random
    feats = compute_features(A, feat_dim, feature_type, rng)
    return A, feats


def sample_ws_graph(n=N_NODES, k=WS_K, p=WS_P, feat_dim=FEAT_DIM, seed=None,
                    feature_type=FEATURE_TYPE):
    # Watts-Strogatz small-world: ring lattice of degree k, edges rewired w.p. p
    G = nx.watts_strogatz_graph(n, k, p, seed=int(seed) if seed is not None else None)
    A = nx.to_numpy_array(G, nodelist=range(n)).astype(np.float32)
    rng   = np.random.RandomState(seed) if seed is not None else np.random
    feats = compute_features(A, feat_dim, feature_type, rng)
    return A, feats


GRAPH_GENERATORS = {
    'ER':  sample_er_graph,
    'RGG': sample_rgg_graph,
    'BA':  sample_ba_graph,
    'WS':  sample_ws_graph,
}


def sample_graph(graph_type, seed=None, **kwargs):
    return GRAPH_GENERATORS[graph_type](seed=seed, **kwargs)


def perturb_graph(A, feats, k, seed=None, feature_type=FEATURE_TYPE):
    rng        = np.random.RandomState(seed) if seed is not None else np.random
    n          = A.shape[0]
    Ap         = A.copy()
    rows, cols = np.triu_indices(n, k=1)
    chosen     = rng.choice(len(rows), min(k, len(rows)), replace=False)
    for idx in chosen:
        r, c = rows[idx], cols[idx]
        Ap[r, c] = Ap[c, r] = 1.0 - Ap[r, c]
    if feature_type == 'degree':
        # recompute degree from the perturbed (input) graph, not the clean one
        featsp = feats.copy()
        featsp[:, 0] = Ap.sum(axis=1) / n
    else:
        # all_ones / random_fixed features are node-level and structure-independent,
        # so they stay unchanged under edge perturbation
        featsp = feats.copy()
    return Ap, featsp


def to_pyg(A, feats, device=DEVICE):
    edge_index = torch.tensor(np.array(np.where(A > 0)), dtype=torch.long)
    x          = torch.tensor(feats, dtype=torch.float32)
    return Data(x=x, edge_index=edge_index, num_nodes=A.shape[0]).to(device)


A, feats   = sample_er_graph(seed=0)
Ap, featsp = perturb_graph(A, feats, k=2, seed=0)
print(f'ER  : {N_NODES} nodes, {int(A.sum()/2)} edges  '
      f'(edit distance after k=2 flip: {int(np.abs(A-Ap).sum()/2)})')

for gt in ['RGG', 'BA', 'WS']:
    Ag, featsg   = sample_graph(gt, seed=0)
    Apg, featspg = perturb_graph(Ag, featsg, k=2, seed=0)
    print(f'{gt:<4}: {N_NODES} nodes, {int(Ag.sum()/2)} edges  '
          f'(edit distance after k=2 flip: {int(np.abs(Ag-Apg).sum()/2)})')
```

## 2. GNN Architectures

**GIN** uses injective sum aggregation — maximally expressive per the WL hierarchy.  
**GCN** uses symmetric normalised mean — smooth but prone to oversmoothing at depth.  
Both L2-normalise embeddings onto the unit hypersphere so distances lie in $[0, 2]$.

```python
class GINEncoder(nn.Module):
    def __init__(self, in_dim, hidden_dim, out_dim, num_layers):
        super().__init__()
        self.convs = nn.ModuleList()
        self.bns   = nn.ModuleList()
        dims = [in_dim] + [hidden_dim] * (num_layers - 1) + [out_dim]
        for i in range(num_layers):
            mlp = nn.Sequential(
                nn.Linear(dims[i], dims[i+1]), nn.ReLU(),
                nn.Linear(dims[i+1], dims[i+1]),
            )
            self.convs.append(GINConv(mlp, train_eps=True))
            self.bns.append(nn.BatchNorm1d(dims[i+1]))

    def forward(self, data):
        x, ei = data.x, data.edge_index
        for conv, bn in zip(self.convs, self.bns):
            x = F.relu(bn(conv(x, ei)))
        return F.normalize(x, p=2, dim=1)


class GCNEncoder(nn.Module):
    def __init__(self, in_dim, hidden_dim, out_dim, num_layers):
        super().__init__()
        self.convs = nn.ModuleList()
        self.bns   = nn.ModuleList()
        dims = [in_dim] + [hidden_dim] * (num_layers - 1) + [out_dim]
        for i in range(num_layers):
            self.convs.append(GCNConv(dims[i], dims[i+1]))
            self.bns.append(nn.BatchNorm1d(dims[i+1]))

    def forward(self, data):
        x, ei = data.x, data.edge_index
        for conv, bn in zip(self.convs, self.bns):
            x = F.relu(bn(conv(x, ei)))
        return F.normalize(x, p=2, dim=1)


ARCH_REGISTRY = {'GIN': GINEncoder, 'GCN': GCNEncoder}


def compute_margin(emb):
    with torch.no_grad():
        d    = torch.cdist(emb, emb, p=2)
        mask = ~torch.eye(d.shape[0], dtype=torch.bool, device=d.device)
        return d[mask].min().item()


def hungarian_alignment_accuracy(emb_A, emb_Ap):
    with torch.no_grad():
        C = torch.cdist(emb_A, emb_Ap, p=2).cpu().numpy()
        if C.max() - C.min() < 1e-8:
            n = C.shape[0]
            return np.mean(np.arange(n) == np.random.permutation(n))
        row_ind, col_ind = linear_sum_assignment(C)
        return np.sum(row_ind == col_ind) / len(row_ind)


for name, Cls in ARCH_REGISTRY.items():
    m   = Cls(FEAT_DIM, HIDDEN_DIM, EMBED_DIM, 3).to(DEVICE)
    out = m(to_pyg(A, feats))
    print(f'{name}: shape={list(out.shape)}  norm={out.norm(dim=1).mean():.4f}  '
          f'margin={compute_margin(out):.4f}')
```

## 3. Triplet Margin Loss

$$\mathcal{L}(m) = \frac{1}{n} \sum_{i=1}^n \max\bigl(0,\; d_{\text{pos}} - d_{\text{neg}} + m\bigr)$$

```python
def triplet_margin_loss(emb_A, emb_Ap, margin):
    n       = emb_A.shape[0]
    d_pos   = (emb_A - emb_Ap).norm(p=2, dim=1)
    shift   = torch.randint(1, n, (n,), device=emb_A.device)
    neg_idx = (torch.arange(n, device=emb_A.device) + shift) % n
    d_neg   = (emb_A - emb_A[neg_idx]).norm(p=2, dim=1)
    return F.relu(d_pos - d_neg + margin).mean()


print('Triplet loss defined.')
```

## 4. Training — Depth × Graph-Family Sweep

Triplet margin $m = 0.5$ fixed. One model per (graph family, depth $L$) pair, i.e. `GRAPH_TYPES` × `DEPTH_VALUES`.  
Architecture is controlled by `ARCH` set in the Setup cell.

```python
def train_model(num_layers, arch=ARCH, margin=FIXED_MARGIN,
                steps=TRAIN_STEPS, seed=42, graph_type='ER'):
    torch.manual_seed(seed); np.random.seed(seed); random.seed(seed)
    Cls   = ARCH_REGISTRY[arch]
    model = Cls(FEAT_DIM, HIDDEN_DIM, EMBED_DIM, num_layers).to(DEVICE)
    opt   = torch.optim.Adam(model.parameters(), lr=LR)
    for _ in range(steps):
        model.train()
        A, feats   = sample_graph(graph_type)
        Ap, featsp = perturb_graph(A, feats, k=0)
        emb_A      = model(to_pyg(A, feats))
        emb_Ap     = model(to_pyg(Ap, featsp))
        loss       = triplet_margin_loss(emb_A, emb_Ap, margin)
        opt.zero_grad(); loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        opt.step()
    return model


print(f'Training {ARCH} depth sweep across graph families {GRAPH_TYPES} (m={FIXED_MARGIN} fixed)...')
depth_models = {gt: {} for gt in GRAPH_TYPES}
for gt in GRAPH_TYPES:
    for L in DEPTH_VALUES:
        print(f'  {gt} L={L} ...', end=' ', flush=True)
        depth_models[gt][L] = train_model(num_layers=L, graph_type=gt)
        print('done')
print('All models trained.')
```

## 5. Evaluation

```python
def evaluate_model(model, k_values=K_VALUES, n_test=N_TEST, graph_type='ER'):
    model.eval()
    gamma_list = []
    acc   = {k: [] for k in k_values}
    shift = {k: [] for k in k_values}
    with torch.no_grad():
        for seed in range(n_test):
            A, feats = sample_graph(graph_type, seed=seed + 2000)
            emb_A    = model(to_pyg(A, feats))
            gamma_list.append(compute_margin(emb_A))
            for k in k_values:
                Ap, featsp = perturb_graph(A, feats, k, seed=seed)
                emb_Ap     = model(to_pyg(Ap, featsp))
                acc[k].append(hungarian_alignment_accuracy(emb_A, emb_Ap))
                shift[k].append(
                    (emb_A - emb_Ap).norm(p=2, dim=1).mean().item()
                )
    return {
        'gamma_mean': np.mean(gamma_list), 'gamma_std': np.std(gamma_list),
        'gamma_all':  gamma_list,
        'acc':        {k: np.mean(v) for k, v in acc.items()},
        'shift':      {k: np.mean(v) for k, v in shift.items()},
        'acc_all': acc, 'shift_all': shift,
    }


print(f'Evaluating {ARCH} depth sweep across graph families {GRAPH_TYPES}...')
depth_results = {gt: {} for gt in GRAPH_TYPES}
for gt in GRAPH_TYPES:
    for L, model in depth_models[gt].items():
        depth_results[gt][L] = evaluate_model(model, graph_type=gt)
        r = depth_results[gt][L]
        print(f'  {gt} L={L}: gamma={r["gamma_mean"]:.4f}+/-{r["gamma_std"]:.4f}  '
              f'acc@k=1={r["acc"][1]:.3f}  acc@k=5={r["acc"][5]:.3f}')
```

## 8. Theory Check: Relative Shift Predicts Accuracy

Proposition 1 gives a sufficient condition: robustness holds when per-node shift $< \hat{\gamma}$.

$$\text{relative shift} = \frac{\|[f(A)]_i - [f(A')]_i\|_2}{\hat{\gamma}_{f(A)}}$$

```python
def collect_relative_shift_records(model, k_values=[1, 2, 3, 5, 8],
                                   n_test=N_TEST, depth=None, graph_type='ER'):
    model.eval()
    records = []
    with torch.no_grad():
        for seed in range(n_test):
            A, feats = sample_graph(graph_type, seed=seed + 2000)
            emb_A    = model(to_pyg(A, feats))
            gamma    = compute_margin(emb_A)
            if gamma < 1e-6:
                continue
            for k in k_values:
                Ap, featsp = perturb_graph(A, feats, k, seed=seed)
                emb_Ap     = model(to_pyg(Ap, featsp))
                mean_shift = (emb_A - emb_Ap).norm(p=2, dim=1).mean().item()
                acc        = hungarian_alignment_accuracy(emb_A, emb_Ap)
                records.append({'k': k, 'gamma': gamma,
                                'rel_shift': mean_shift / gamma,
                                'acc': acc, 'depth': depth, 'graph_type': graph_type})
    return records


all_records = []
for gt in GRAPH_TYPES:
    for L, model in depth_models[gt].items():
        all_records.extend(collect_relative_shift_records(model, depth=L, graph_type=gt))

rel_shifts = np.array([r['rel_shift']  for r in all_records])
accs_arr   = np.array([r['acc']        for r in all_records])
depths_arr = np.array([r['depth']      for r in all_records])
gtypes_arr = np.array([r['graph_type'] for r in all_records])

bins = np.unique(np.percentile(rel_shifts, np.linspace(0, 100, 21)))
bin_mid, bin_acc, bin_std = [], [], []
for lo, hi in zip(bins[:-1], bins[1:]):
    mask = (rel_shifts >= lo) & (rel_shifts < hi)
    if mask.sum() > 3:
        bin_mid.append((lo + hi) / 2)
        bin_acc.append(accs_arr[mask].mean())
        bin_std.append(accs_arr[mask].std())

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
fig.suptitle(f'Theory Check: Relative Shift vs. Accuracy ({ARCH}), pooled over {", ".join(GRAPH_TYPES)}',
             fontsize=12, fontweight='bold')

ax = axes[0]
for gt in GRAPH_TYPES:
    mask = gtypes_arr == gt
    ax.scatter(rel_shifts[mask], accs_arr[mask], alpha=0.3, s=12,
               color=GRAPH_COLORS[gt], label=gt)
ax.axvline(1.0, color='red', linestyle='--', linewidth=1.8,
           label='shift $= \\hat{\\gamma}$')
ax.set_xlabel('Relative shift  $\\|f(A)-f(A^\\prime)\\| / \\hat{\\gamma}_{f(A)}$')
ax.set_ylabel('Alignment accuracy')
ax.legend(fontsize=9, markerscale=2, title='graph family')

ax = axes[1]
ax.errorbar(bin_mid, bin_acc, yerr=bin_std, fmt='o-', color='steelblue',
            capsize=3, linewidth=2, markersize=6)
ax.axvline(1.0, color='red', linestyle='--', linewidth=1.5,
           label='shift $= \\hat{\\gamma}$')
ax.axhline(0.9, color='grey', linestyle=':', linewidth=1.0, label='90% cutoff')
ax.set_xlabel('Relative shift  $\\|f(A)-f(A^\\prime)\\| / \\hat{\\gamma}_{f(A)}$')
ax.set_ylabel('Mean alignment accuracy (pooled over graph families)')
ax.legend(fontsize=9)

plt.tight_layout()
plt.savefig(f'ALL_{ARCH.lower()}_depth_theory_check.pdf', bbox_inches='tight')
plt.savefig(f'ALL_{ARCH.lower()}_depth_theory_check.png', bbox_inches='tight', dpi=150)
plt.show()
print(f'Theory check complete. Total data points: {len(all_records)} across {GRAPH_TYPES}.')
```
